```python
import os
import glob
import numpy as np
from skimage.segmentation import slic, mark_boundaries
from skimage.util import img_as_float
from skimage import io
import matplotlib.pyplot as plt

# --- 1. 設定您的路徑和參數 ---

# 輸入的圖片資料夾
INPUT_DIR = '/home/U116med/data/polyp/TrainDataset/images' 
# 輸出的 pseudolabel 儲存資料夾
OUTPUT_DIR = './polypdata/pseudolabels/'
# 輸出的視覺化結果儲存資料夾 (強烈建議保留，用於檢查結果)
VISUALIZATION_DIR = './polypdata/visualizations/'

# SLIC 演算法參數
# n_segments: 您希望將圖片分割成大約多少個 superpixel
N_SEGMENTS = 300
# compactness: 平衡顏色和空間的權重。值越大，superpixel 形狀越方正。
COMPACTNESS = 10
# sigma: 在計算 SLIC 前，對圖片進行高斯模糊的標準差，可以讓邊界更平滑
SIGMA = 1


def generate_pseudolabels_for_2d_color(image_path, output_dir, vis_dir):
    """
    為單張2D彩色圖片生成superpixel pseudolabels。

    Args:
        image_path (str): 輸入圖片的路徑。
        output_dir (str): 儲存pseudolabel的路徑。
        vis_dir (str): 儲存視覺化結果的路徑。
    """
    # 讀取圖片
    try:
        image = io.imread(image_path)
    except Exception as e:
        print(f"Error reading {image_path}: {e}")
        return

    # 確保圖片是RGB格式 (移除可能存在的Alpha通道)
    if image.shape[-1] == 4:
        image = image[..., :3]
        
    # 將圖片轉換為float格式，這是skimage演算法的標準輸入
    image_float = img_as_float(image)

    # --- 2. 使用 SLIC 演算法生成 Superpixels ---
    # segments 是一個 2D 陣列，每個 pixel 的值代表它所屬的 superpixel ID
    # 例如，所有值為 10 的 pixel 組成了第10個 superpixel
    segments = slic(image_float, 
                    n_segments=N_SEGMENTS, 
                    compactness=COMPACTNESS, 
                    sigma=SIGMA,
                    start_label=1) # 讓標籤從1開始，避免0被誤認為背景

    # 取得檔案基本名稱，用於儲存
    base_name = os.path.splitext(os.path.basename(image_path))[0]
    
    # --- 3. 儲存 Pseudolabel 結果 ---
    # pseudolabel 是一個整數標籤圖 (integer label map)
    # 我們將它儲存為16位元的PNG，以支援超過255個標籤
    output_filename = os.path.join(output_dir, f"{base_name}_label.png")
    # 注意：直接打開這個PNG檔案可能會看到全黑，因為標籤值(1, 2, 3...)很小。
    # 但資料是正確的，程式可以正常讀取。
    io.imsave(output_filename, segments.astype(np.uint16))

    # --- 4. 產生並儲存視覺化結果 ---
    # 這一步驟非常重要，可以讓您直觀地檢查 superpixel 的分割效果
    vis_image = mark_boundaries(image_float, segments)
    vis_filename = os.path.join(vis_dir, f"{base_name}_vis.png")
    io.imsave(vis_filename, (vis_image * 255).astype(np.uint8))
    
    print(f"Processed {base_name} successfully.")


def main():
    # 建立輸出資料夾
    os.makedirs(OUTPUT_DIR, exist_ok=True)
    os.makedirs(VISUALIZATION_DIR, exist_ok=True)

    # 找到所有PNG圖片
    image_paths = glob.glob(os.path.join(INPUT_DIR, '*.png'))
    # 如果您的圖片是 jpg 或其他格式，請修改這裡
    # image_paths.extend(glob.glob(os.path.join(INPUT_DIR, '*.jpg')))
    
    if not image_paths:
        print(f"Error: No images found in {INPUT_DIR}. Please check the path.")
        return

    print(f"Found {len(image_paths)} images. Starting processing...")
    
    # 遍歷所有圖片並處理
    for path in image_paths:
        generate_pseudolabels_for_2d_color(path, OUTPUT_DIR, VISUALIZATION_DIR)
        
    print("All images processed.")


if __name__ == '__main__':
    main()
```
