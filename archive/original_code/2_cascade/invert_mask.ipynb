```python
import cv2
import os
import numpy as np
from tqdm import tqdm

def invert_mask(input_path, output_path):
    """
    讀取 Mask，將前景變為黑色 (0)，背景變為白色 (255)
    """
    # 1. 以灰階模式讀取
    mask = cv2.imread(input_path, cv2.IMREAD_GRAYSCALE)
    
    if mask is None:
        print(f"Error: 無法讀取圖片 {input_path}")
        return

    # 2. 二值化 (Thresholding) - 確保圖片只有純黑與純白
    # 假設原圖前景是亮色(>127)，背景是暗色
    # ret, binary = cv2.threshold(mask, 127, 255, cv2.THRESH_BINARY)
    
    # 如果你的原圖已經是很好的 0/255，可以直接用 mask，但建議做一步二值化以防萬一
    # 這裡假設原圖：前景=255(白), 背景=0(黑)
    _, binary = cv2.threshold(mask, 127, 255, cv2.THRESH_BINARY)

    # 3. 反轉顏色 (Invert)
    #原本: 前景(255), 背景(0) -> bitwise_not -> 前景(0), 背景(255)
    inverted_mask = cv2.bitwise_not(binary)

    # 4. 存檔
    cv2.imwrite(output_path, inverted_mask)

def process_folder(input_dir, output_dir):
    """
    批量處理整個資料夾
    """
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
        
    files = [f for f in os.listdir(input_dir) if f.endswith(('.png', '.jpg', '.jpeg', '.bmp'))]
    
    print(f"正在處理 {len(files)} 張圖片...")
    for filename in tqdm(files):
        in_path = os.path.join(input_dir, filename)
        out_path = os.path.join(output_dir, filename)
        invert_mask(in_path, out_path)
    print("完成！")

if __name__ == '__main__':
    # --- 設定路徑 ---
    
    # 模式 A: 單張測試
    input_img = '/home/U116med/wch_code/cascade/check_functionality_data/masks/110.png'
    output_img = '110_mask_inverted.png'
    invert_mask(input_img, output_img)
    
    # 模式 B: 批量處理 (比如你的測試集 Masks)
    #input_folder = '/home/U116med/data/polyp/TestDataset/test/masks/'  # 修改為你的輸入資料夾
    #output_folder = './inverted_masks_output/'                         # 修改為你的輸出資料夾
    
    #process_folder(input_folder, output_folder)
```
